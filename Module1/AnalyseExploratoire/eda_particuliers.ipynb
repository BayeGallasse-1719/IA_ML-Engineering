```python
## separation nom variable qualitative et quantitative
vars_quali = df.select_dtypes(exclude=np.number).columns.tolist()
vars_quanti = df.select_dtypes(include=np.number).columns.tolist()

## variables quantitative
print(df[vars_quanti].describe())
print()

## variable qualitative
print(df[vars_quali].describe().T)
```

```python
## Analyse des variables qualitative qui devait etre quantitative
for var in vars_quali:
    print(f"{var:-<40}{list(df[var].unique())}")
```
